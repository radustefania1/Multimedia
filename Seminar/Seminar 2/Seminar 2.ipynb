#### versiunea 1

```python
penguins.loc[penguins['body_mass_g'] < 4000, 'sex'] = 'Female'
penguins.loc[penguins['body_mass_g'] >= 4000, 'sex'] = 'Male'
penguins.isna().sum()
```

// apare ca doi pinguini nu au sex pentru ca nu au greutate, deci nu au cum sa compare cu 4000

```python
penguins[penguins['sex'].isna()]
```

#### versiunea 2

```python
penguins = sns.load_dataset("penguins")
penguins["sex"] = np.where(
    penguins["sex"].isna(),
    np.where(penguins["body_mass_g"]<4000,
            "Female",
            np.where(penguins["body_mass_g"]>= 4000,
                     "Male",
                     None)),
    penguins["sex"])

penguins.head()
```

// a doua e mai precisa, la prima imi punea la un pinguin male doar pentru ca nu avea greutatea mai mica de 4000, chiar daca pinguinul ala nu avea greutate

```python
penguins = penguins.dropna()

penguins.isna().sum()
```

```python
penguins["bill_area"] = penguins["bill_length_mm"] * penguins["bill_depth_mm"]
penguins.head()
```

```python
penguins[["species","bill_length_mm","bill_depth_mm", "flipper_length_mm","bill_area"]].groupby("species").aggregate("mean")
```
